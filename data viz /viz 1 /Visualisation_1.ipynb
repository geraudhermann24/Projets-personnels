## SUJET I: [NUTRITION] MODE DE CONSOMMATION DANS LE MONDE

Importation des données

```python
url =  "https://raw.githubusercontent.com/She-Codes-Now/Intro-to-Data-Science-with-R/master/food_coded.csv"

df_food = pd.read_csv(url)
df_food.head()
```

```python
df_food.shape
```

Repartition selon le genre(sexe)

```python
val_nul=df_food.isnull().sum().sum
print("Le nombre de NaN trouvé est: ", val_nul)
```

```python
#Enlever toutes les colonnes qui ont plus de 1à Nan
df_food_NA=df_food.loc[:,df_food.isnull().sum()<=10]
df_food_NA.dropna(inplace=True)
df_food_NA.shape
```

```python
df_food_NA.columns
```

```python
food_var_global=df_food_NA.loc[:,('GPA', 'Gender', 'father_profession', 'fav_cuisine','mother_education', 'mother_profession',
       'nutritional_check','parents_cook', 'pay_meal_out')]
food_var_global.shape
```

```python
#-------------------Préparation des données---------------------------
counts=food_var_global['Gender'].value_counts()
counts = counts.rename({1: 'Femme', 2: 'Homme'})
#-------------------Graphique sur les données---------------------------
plt.pie(counts.values, labels=counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Répartition en fonction du genre')
plt.legend()
plt.show()
```

```python
#food_var_global.Gender.rename({1: 'Femme', 2: 'Homme'}, inplace=True)
food_var_global['GPA'] = pd.to_numeric(food_var_global['GPA'], errors='coerce')

food_var_global['GPA'] = food_var_global['GPA'].round().astype('Int64')  # Int64 accepte les NaN


plt.figure(figsize=(6,4))
plt.scatter(x='Gender', y='GPA', data=food_var_global)
plt.xlabel('Genre')
plt.ylabel('GPA')
plt.title('Distribution des GPA par genre')
#plt.xticks([0,1], ['Femme', 'Homme'])  # si besoin, adapte selon codage Gender
plt.show()

```
